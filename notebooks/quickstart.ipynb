## CSR problem statement

## Headless browser

```python
# Create and start browser manager
manager = BrowserManager()
await manager.start()
```

```python
# Create fetcher and fetch the page
async with BrowserFetcher(manager) as fetcher:
    page = await fetcher.fetch(URL)

page.status, page.title
```
